```python
import pandas as pd
import numpy as np
import gzip
import pickle
import sys
import os

caminho_pai = os.path.abspath(os.path.join(os.getcwd(), os.pardir))
if caminho_pai not in sys.path:
    sys.path.append(caminho_pai)

caminho_interno = os.path.join(caminho_pai, "numpy_model")
if caminho_interno not in sys.path:
    sys.path.append(caminho_interno)

from numpy_model.data import Data

# ==========================================
# 1. CARREGAR FICHEIROS E MODELO
# ==========================================
print("A carregar modelo e dados...")

df_subm = pd.read_csv('subm1.csv', sep=';')


# Carregar o vocabulário salvo do treino
with open('vocab_numpy.pkl', 'rb') as f:
    vocab = pickle.load(f)

# Carregar o modelo treinado (obtido no treino do m1-NeuralNetwork.ipynb)
with gzip.open('modelo_numpy_final.pkl.gz', 'rb') as f:
    modelo = pickle.load(f)

# Recriar o dicionário para a vetorização
word_to_idx = {palavra: i for i, palavra in enumerate(vocab)}

# ==========================================
# 2. TRANSFORMAR O TEXTO EM NÚMEROS
# ==========================================
print("A vetorizar os textos do subm1.csv...")
def vetorizar_texto(texto):
    vetor = np.zeros(len(vocab))
    for palavra in str(texto).lower().split():
        if palavra in word_to_idx:
            vetor[word_to_idx[palavra]] += 1
    soma = np.sum(vetor)
    if soma > 0:
        vetor = vetor / soma
    return vetor

# Criar o X (matriz de textos)
X_subm = np.array([vetorizar_texto(t) for t in df_subm['Text']])

# Criar um 'y' falso (só de zeros) porque a nossa classe 'Data' obriga a ter um y
y_dummy = np.zeros((len(X_subm), 5)) 
dataset_subm = Data(X=X_subm, y=y_dummy)

# ==========================================
# 3. FAZER A CLASSIFICAÇÃO
# ==========================================
print("A classificar (IAs vs Humano)...")
previsoes_prob = modelo.predict(dataset_subm)

# Descobrir qual a classe com maior probabilidade
classes_previstas_idx = np.argmax(previsoes_prob, axis=1)

idx_to_label = {0: 'Human', 1: 'OpenAI', 2: 'Google', 3: 'Meta', 4: 'Anthropic'}

# Criar a coluna final "Labels" com os nomes
df_subm['Labels'] = [idx_to_label[idx] for idx in classes_previstas_idx]

# ==========================================
# 4. EXPORTAR O FICHEIRO FINAL
# ========================================== 

nome_ficheiro_saida = f"subm1-g7-MEI-A.csv"

# Guardar o CSV
df_subm.to_csv(nome_ficheiro_saida, sep=';', index=False)

print(f"Concluído!")
# Visualizar as primeiras linhas
df_subm.head()
```
