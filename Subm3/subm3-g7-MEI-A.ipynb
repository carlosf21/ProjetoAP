```python
import pandas as pd
import torch
from transformers import pipeline

# ==========================================
# 1. CONFIGURAÇÃO DO MODELO
# ==========================================
nome_do_modelo = "carlosf21/projeto-ap-modernbert"

print(f"A descarregar o modelo '{nome_do_modelo}'... ")
device = 0 if torch.cuda.is_available() else -1

# Criar a pipeline (ela descarrega os pesos e o tokenizer automaticamente)
classificador = pipeline(
    "text-classification", 
    model=nome_do_modelo, 
    device=device
)
```

```python
# ==========================================
# 2. CARREGAR O FICHEIRO DE TESTE
# ==========================================
print("A ler o ficheiro 'subm3.csv'...")
df_teste = pd.read_csv('subm3.csv', sep=';')
```

```python
# ==========================================
# 3. FAZER AS PREVISÕES
# ==========================================
print("A analisar os textos e a gerar previsões...")
# Passamos a lista toda de uma vez. O truncation=True e max_length=512 garantem que textos gigantes não crasham o modelo.
resultados = classificador(df_teste['Text'].tolist(), truncation=True, max_length=512)

# O pipeline devolve algo como: [{'label': 'Human', 'score': 0.98}, {'label': 'OpenAI', 'score': 0.95}]
# Vamos extrair apenas o valor da 'label'
print(resultados[:5])  # Mostrar os primeiros 5 resultados para verificar
df_teste['Label'] = [res['label'] for res in resultados]
print(classificador.model.config.id2label)
```

```python
# ==========================================
# 4. EXPORTAR O FICHEIRO FINAL
# ==========================================
nome_ficheiro_final = 'subm3-g7-MEI-A.csv'

df_teste[['ID', 'Label']].to_csv(nome_ficheiro_final, sep=';', index=False)

print(f"O ficheiro '{nome_ficheiro_final}' foi gerado com sucesso.")
```
